```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_groq import ChatGroq


# 2. Prompt Template (Memory placeholder ke sath)
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "Tum ek madadgar AI assistant ho.roman urdu may jawab do"),
        MessagesPlaceholder(variable_name="history"),
        ("human", "{input}"),
    ]
)

# 3. Chain
chain = prompt | llm

# 4. Simple Memory Store
store = {}


def get_session_history(session_id: str):
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]


# 5. Conversation Chain (New Replacement)
conversation = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="input",
    history_messages_key="history",
)

config = {"configurable": {"session_id": "session_1"}}

# Conversation
print("Turn 1:")
res1 = conversation.invoke({"input": "Mera naam Noman hai"}, config=config)
print(res1.content)

print("\nTurn 2:")
res2 = conversation.invoke({"input": "Mera naam kya hai?"}, config=config)
print(res2.content)
```

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_groq import ChatGroq

# 1. LLM Setup
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    api_key="GROQ_API_KEY",  # Apni API key yahan likhein
)

# 2. Prompt Template
prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "Tum PakShop ke helpful assistant ho. Roman Urdu mein jawab do.",
        ),
        MessagesPlaceholder(variable_name="chat_history"),
        ("human", "{user_query}"),
    ]
)

# 3. Chain
chain = prompt | llm

# 4. Memory Register (Store)
store = {}


def get_history(session_id: str):
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]


# 5. Manager (RunnableWithMessageHistory)
bot = RunnableWithMessageHistory(
    chain,
    get_history,
    input_messages_key="user_query",  # Is variable name par dhyan dein
    history_messages_key="chat_history",  # Is variable name par dhyan dein
)

# --- USER 1: NOMAN ---
config_noman = {"configurable": {"session_id": "session_noman"}}

bot.invoke(
    {"user_query": "Mera naam Noman hai aur mujhe Laptop chahiye."},
    config=config_noman,
)

# --- USER 2: ALI ---
config_ali = {"configurable": {"session_id": "session_ali"}}

bot.invoke(
    {"user_query": "Mera naam Ali hai aur mujhe Mobile chahiye."},
    config=config_ali,
) 

# --- TESTING MEMORY ---
print("--- Noman Ka Response ---")
res_noman = bot.invoke(
    {"user_query": "Mujhe kya khareedna tha?"}, config=config_noman
)
print(res_noman.content)

print("\n--- Ali Ka Response ---")
res_ali = bot.invoke(
    {"user_query": "Mujhe kya khareedna tha?"}, config=config_ali
)
print(res_ali.content)
```

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    api_key="your_api_key_here",  # Apni API key yahan likhein
)

prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "Tum PakShop ke helpful assistant ho. Roman Urdu mein jawab do.",
        ),
        MessagesPlaceholder(variable_name="chat_history"),
        ("human", "{user_query}"),
    ]
)

chain = prompt | llm

store = {}

def get_history(session_id: str):
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]

bot = RunnableWithMessageHistory(
    chain,  
    get_history,
    input_messages_key="user_query",
    history_messages_key="chat_history",
)

config_noman  = {"configurable" : {"session_id" : "session_noman"}}
bot.invoke(
    {"user_query": "Mera naam Noman hai aur mujhe Laptop chahiye."},
    config=config_noman,
)

config_ali = {"configurable" : {"session_id" : "session_ali"}}
bot.invoke(
    {"user_query": "Mera naam Ali hai aur mujhe Mobile chahiye."},
    config=config_ali,
)

print("--- Noman Ka Response ---")
res_noman = bot.invoke(
    {"user_query": "Mujhe kya khareedna tha?"}, config=config_noman
)
print(res_noman.content)
print("\n--- Ali Ka Response ---")
res_ali = bot.invoke(
    {"user_query": "Mujhe kya khareedna tha?"}, config=config_ali
)
print(res_ali.content)
```
