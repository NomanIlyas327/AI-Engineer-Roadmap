```python
from groq import Groq

client = Groq(api_key="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")

# System Message Dictionary Structure Mein
message_history = [
    {
        "role": "system",
        "content": "You are a helpful Python Assistant. Always respond in simple roman Urdu with short examples."
    }
]

print("--- Chatbot Started ---")

while True:
    input_message = input("\nAap: ")
    
    if input_message.lower() in ["quit", "exit"]:
        print("Shukriya! Allah Hafiz!")
        break
        
    # User message append karein
    message_history.append({"role": "user", "content": input_message})
    
    # API Call
    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=message_history,
        stream=True
    )
    
    print("Assistant: ", end="", flush=True)
    full_response = ""
    
    # Streaming Loop
    for chunk in response:
        # Check karein ke text chunk mein content mojood hai ya nahi
        if chunk.choices and chunk.choices[0].delta.content:
            text = chunk.choices[0].delta.content
            print(text, end="", flush=True)
            full_response += text
            
    print() # New line after completion
    
    # AI ka response SIRF DICTIONARY format mein history mein daalein
    message_history.append({"role": "assistant", "content": full_response})
```

```python
from groq import Groq
import json

# Function Definition
def get_current_weather(location):
    if "lahore" in location.lower():
        return json.dumps({"location": "Lahore", "temperature": "32°C", "condition": "Sunny"})
    else:
        return json.dumps({"location": location, "temperature": "25°C", "condition": "Cloudy"})

# Tools Schema Definition
tools = [
    {
        "type": "function",
        "function": {
            "name": "get_current_weather",
            "description": "Get the current weather for a given city",
            "parameters": {
                "type": "object",
                "properties": {
                    "location": {
                        "type": "string",
                        "description": "The city and state, e.g. Lahore, Pakistan",
                    }
                },
                "required": ["location"],
            },
        },
    }
]

# Client setup
client = Groq(api_key="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")

messages = []

print("--- AI Weather Agent Started ---")

while True:
    user_input = input("\nAap: ")
    if user_input.lower() in ["quit", "exit"]:
        print("Shukriya! Allah Hafiz!")
        break
        
    messages.append({"role": "user", "content": user_input})

    # API Call Loop Ke ANDAR Honi Chahiye
    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=messages,
        tools=tools,
        tool_choice="auto"
    )

    response_message = response.choices[0].message

    # Check karein ke kya AI ne Tool/Function Call decide kiya hai?
    if response_message.tool_calls:
        print("\n🤖 AI ne Function Call Detect Kiya!")
        tool_call = response_message.tool_calls[0]
        
        # Function name aur extracted arguments print karein
        function_name = tool_call.function.name
        function_args = tool_call.function.arguments
        
        print(f"Function Name: {function_name}")
        print(f"Extracted Arguments (JSON): {function_args}")
        
    else:
        # Agar AI normal message de raha hai
        print(f"\nAssistant: {response_message.content}")
```
