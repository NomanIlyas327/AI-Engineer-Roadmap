```python
np.random.seed(42)
n = 1000

# ✅ Extra commas hata diye hain:
Tenure_Months = np.random.randint(1, 60, n)
Monthly_GB_Used = np.random.randint(10, 500, n)
Customer_Service_Calls = np.random.randint(0, 8, n)
Contract_Type = np.random.choice(['Monthly', 'Yearly'], size=n)
Noise = np.random.normal(0, 5, n)

# Target 1: Regression
Monthly_Bill = 20 + (Tenure_Months * 0.5) + (Monthly_GB_Used * 0.2) + Noise

# Target 2: Classification
Churn = np.where(Customer_Service_Calls > 3, 1, 0)

# DataFrame Creation
df = pd.DataFrame({
    'Tenure_Months': Tenure_Months,
    'Monthly_GB_Used': Monthly_GB_Used,
    'Customer_Service_Calls': Customer_Service_Calls,
    'Contract_Type': Contract_Type,
    'Monthly_Bill': Monthly_Bill,
    'Churn': Churn
})


# --- CELL 2: DATA EXPLORATION ---
print("--- FIRST 5 ROWS ---")
display(df.head())

print("\n--- MISSING VALUES ---")
print(df.isnull().sum())

print("\n--- SUMMARY STATISTICS ---")
display(df.describe())
```

```python
from sklearn.model_selection import train_test_split


df.to_csv('synthetic_telecom_data.csv', index=False)
print("\nSynthetic dataset saved as 'synthetic_telecom_data.csv'.")

df = pd.read_csv('synthetic_telecom_data.csv')

df_encoded = pd.get_dummies(df, columns=['Contract_Type'], drop_first=True)

X = df_encoded.drop(['Monthly_Bill', 'Churn'], axis=1)
y_reg = df_encoded['Monthly_Bill']
y_clf = df_encoded['Churn']
# 1. Regression Split
X_train_reg, X_test_reg, y_reg_train, y_reg_test = train_test_split(
    X, y_reg, test_size=0.2, random_state=42
)

# 2. Classification Split (With Stratification)
X_train_cls, X_test_cls, y_clf_train, y_clf_test = train_test_split(
    X, y_clf, test_size=0.2, random_state=42, stratify=y_clf
)

print("--- SPLIT SUCCESSFUL ---")
print("Regression Training Shape    :", X_train_reg.shape)  # (800, 4)
print("Regression Testing Shape     :", X_test_reg.shape)   # (200, 4)
print("Classification Training Shape:", X_train_cls.shape)  # (800, 4)
print("Classification Testing Shape :", X_test_cls.shape)   # (200, 4)
```
